The **LeakyVault Bank** offers several services to its customers. One of those services is mortage lending. The **LeakyVault Bank** carries out a risk study to decide if it should approve the loan for a customer. To keep it simple let's assume that the risk study is to check that customer salary is bigger than 60000 € per year. But **LeakyVault Bank** does not want customers to know the minimum salary it requires to approve a loan.

To check that one customer has a bigger salary than 60000 € the bank will encrypt -60000 using the Paillier algorithm. Then the crypted text for -60000 will be sent to the customer. Customer will not be able to decrypt it due to it was encrypted using the **LeakyVault Bank** key. So privacy is granted to the **LeakyVault Bank**. Customer will add its salary to the crypted text and send the result back to the **LeakyVault Bank**. The **LeakyVault Bank** will decrypt the result and if the result is bigger than 0 that means that customer salary is bigger than 60000 €.

The first step is to create bank's keys:

```python
from phe import paillier

# Bank create its keys
bank_public_key, bank_private_key = paillier.generate_paillier_keypair()
```

The bank ciphers the requeriment, the $60000:

```python
# The bank encrypts the requeriments, it encrypts it as a negative number due to the bank will ask the customer to add it salary
requeriment = -60000
crypted_requeriment = bank_public_key.encrypt(requeriment)
```

We can check the **crypted_requeriment** variable:

```python
crypted_requeriment
```

We can see that it is not a number.

The bank sends **crypted_requeriment** to the customer applying for the mortage, requesting the client to add its salary to it:

```python
bank_requeriment = 60001 + crypted_requeriment
```

We can check the raw variable to see that it is not a number:

```python
bank_requeriment
```

Client send **band_requeriment** to the bank and bank decrypts it:

```python
result = bank_private_key.decrypt(bank_requeriment)

if result > 0:
    print('Customer salary per year is bigger than 60000 €, so loan will be granted (Result: %d).' % result)
else:
    print('Customer salary per year is shorter than 60000 €, so loan will not be granted (Result: %d).' % result)
```

With this simple example we have seen:

* How the **LeakyVault Bank** can keep its requirement of 60000 €/year secret.
* Customer does not need to access the **LeakyVault Bank** keys to encrypt the data.
