```python
import numpy as np
from concrete import fhe

# Función que queremos evaluar homomórficamente (máximo entre dos enteros)
def encrypted_max(a: int, b: int) -> int:
    delta = a - b
    pos = np.maximum(delta, 0)  # max(0, delta)
    return b + pos

# Compilamos la función para que Concrete genere el circuito homomórfico
compiler = fhe.Compiler(encrypted_max, {"a": "encrypted", "b": "encrypted"})

# Conjunto de ejemplos para ayudar a determinar los rangos y optimizar el circuito
inputset = [(5, 3), (2, 7), (9, 4), (8, 8), (0, 10)]
circuit = compiler.compile(inputset)

# Generación de claves (incluye claves para bootstrapping)
circuit.keygen()

# Cifrado de dos valores de entrada
ct_a, ct_b = circuit.encrypt(6, 4)

# Evaluación homomórfica que incluye bootstrapping internamente
ct_max = circuit.run(ct_a, ct_b)

# Descifrado del resultado
res = circuit.decrypt(ct_max)
print("Máximo de 6 y 4 (homomórfico):", res)
```
